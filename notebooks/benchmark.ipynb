## Seleccionar modelos a comparar

```python
# Pequeños
models = ["microsoft/Phi-3-mini-4k-instruct", "google/gemma-2b", "meta-llama/Llama-3.2-1B", "bigcode/starcoder2-3b"]

# Medianos
models = ["meta-llama/Llama-3.1-8B-Instruct", "mistralai/Mistral-7B-v0.1", "Qwen/Qwen2.5-7B-Instruct", "BioMistral/BioMistral-7B"]

#Grandes
models = ["meta-llama/Llama-3.3-70B-Instruct", "mistralai/Mixtral-8x22B-Instruct-v0.1", "Qwen/Qwen2.5-72B-Instruct", "seliny2/Medusa-Llama-3.1-70B"]
```

## Seleccionar rutas a los prompts

```python
prompts_paths = ["greenllm/data/prompts/prompt_set_small.txt", "greenllm/data/prompts/prompt_set_medium.txt"]
for prompts_path in prompts_paths:
  print(prompts_path)
```

## Seleccionar categorías de conocimiento

```python
categories = "all"
```

## Ejecutar benchmark

```python
from framework.src.greenllm.runners.inference_runner import run_benchmark

results = run_benchmark(model_names=models, prompts_paths=prompts_paths, max_new_tokens=64, categories = categories)


```

## Obtener métricas

```python
from framework.src.greenllm.metrics.metrics import compute_benchmark_metrics

metrics = compute_benchmark_metrics(results)


```
